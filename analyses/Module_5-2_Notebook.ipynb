## Part 1: EDA on Births Data
The following code performs exploratory data analysis (EDA) on the 1971 birth data from Module 4.

```python
from pathlib import Path
import pandas as pd

# The notebook is stored in the repository's analyses/ folder.
# Go up one level to the base repository folder, then into data/ and results/.
base_dir = Path("..").resolve()
data_dir = base_dir / "data"
results_dir = base_dir / "results"

# Create the results directory if it does not already exist.
results_dir.mkdir(parents=True, exist_ok=True)

# Original data source URL, kept for reference only.
births_url = "https://github.com/DAIR3/DAIR3-Workshop/raw/refs/heads/main/resources/unit_4/data/1971.csv.gz"

# Data files are assumed to already exist in ../data.
births_filename = data_dir / "1971.csv.gz"
```

```python
file_path = births_filename

if file_path.is_file():
    print(f"The file exists: {file_path}")
else:
    print(f"File not found: {file_path}")
```

```python
births = pd.read_csv(births_filename)
births[0:5]
```

```python
births.shape
```

```python
births.columns
```

```python
import matplotlib.pyplot as plt

data = births['momage']

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color='skyblue', edgecolor='black')

plt.title('Histogram of Mother Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.tight_layout()

plot_path = results_dir / "mother_ages_histogram.png"
plt.savefig(plot_path, dpi=300, bbox_inches='tight')
print(f"Plot saved to {plot_path}")

plt.show()
```

```python
birthorder_freqs = births['birthorder'].value_counts().reset_index(name='count').sort_values(by='birthorder')
fancy_table_style = birthorder_freqs.style.background_gradient(cmap='Blues')

birthorder_freqs_path = results_dir / "birthorder_freqs.html"
fancy_table_style.to_html(birthorder_freqs_path)
print(f"Birth order frequency table saved to {birthorder_freqs_path}")

fancy_table_style
```
